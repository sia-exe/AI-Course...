```python
place = input("Enter your travel destination: ")

daily_cost = {
    "dubai": 15000,
    "turkey": 12000,
    "malaysia": 9000
 }

print("\n Agent processing...")

if place in daily_cost:
    days = int(input("How many days do you want to stay? "))

    rate = daily_cost[place]
    total_cost = rate * days

    print("\nAgent: Your estimated trip cost is", total_cost, "PKR")

    if days <= 7:
        print("Agent: This is a short and suitable trip.")
    else:
        print("Agent: This is a long trip. Plan your expenses carefully.")
else:
    print("Agent: Sorry, this destination is not available.")
```
